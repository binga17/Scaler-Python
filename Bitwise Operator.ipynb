```python
#Bitwise AND (&) Sets each bit 1 if both bits are 1
a = 5  #0101 in binary
b = 3  #0011 in binary
print(a & b)
```

```python
#Bitwise OR (|) Sets each bit to 1 if at least one of the bits is 1 
a = 5 #0101
b = 3 #0011
print(a | b) 
```

```python
#Bitwise XOR(^) Sets each bit to 1 if only one of the bits is 1
a = 5  # 0101
b = 3  # 0011
print(a ^ b)
```

```python
#Bitwise Not(~) Flips all the bits
a = 5  # 0101
print(~a)
```

```python
#Left Shift (<<) 
a = 5  # 0101
print(a << 1) 
```

```python
#Right Shift (>>)
a = 5  #0101
print(a >> 1)
```

```python
# Identity Operators
a = 5
b = 5
print(a is b)

#it works only for variable not for the complex data types like lists, tuples, strings
```

```python
#Membership Operators
lst = [1, 2, 3, 4]
print(1 in lst)
```

```python
d = {1: "a", 2: "b"} #can be checked only for keys not for values
print(1 in d)
```

```python
#In membership operators, when the values are presented through 
#dictionary, you can only check for key but not for values
print("a" in d) 
```
