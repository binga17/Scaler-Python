```python
#Bitwise AND (&) Sets each bit 1 if both bits are 1
a = 5  #0101 in binary
b = 3  #0011 in binary
print(a & b)
```

```python
#Bitwise OR (|) Sets each bit to 1 if at least one of the bits is 1 
a = 5 #0101
b = 3 #0011
print(a | b) 
```

```python
#Bitwise XOR(^) Sets each bit to 1 if only one of the bits is 1
a = 5  # 0101
b = 3  # 0011
print(a ^ b)
```

```python
#Bitwise Not(~) Flips all the bits
a = 5  # 0101
print(~a)
```

```python
#Left Shift (<<) 
a = 5  # 0101
print(a << 1) 
```

```python
#Right Shift (>>)
a = 5  #0101
print(a >> 1)
```
