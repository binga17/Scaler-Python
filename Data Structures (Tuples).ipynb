```python
#tuple creation
#empty tuple
t = ()

#tuple having integers
t = (1, 2, 3)
print(t)

#tuple with mixed datatypes
t = (1, 'raju', 28, 'abc')
print(t)

#nested tuple
t = (1, (2, 3, 4), [1, 'raju', 28, 'abc'])
print(t)
                
```

```python
#only paranthesis is not enough
t = ('sathish')
type(t)
```

```python
#need a comma at the end
t = ('sathish',)
type(t)
```

```python
#paranthesis id optional
t = 'sathish',
print(type(t))

print(t)
```

```python
#accessing elements in tuple
t = ('satish', 'murali', 'naveen', 'srinu', 'brahma')

print(t[1])
```

```python
#negative index
print(t[-1])
```

```python
#nested tuple
t = ('ABC', ('satish', 'naveen', 'srinu'))

print(t[1])
```

```python
print(t[1][2])
```

```python
#slicing
t = (1, 2, 3, 4, 5, 6)

print(t[1:4])

#print elements from starting to 2nd last elements
print(t[:-2])

#print elements from starting to end
print(t[:])
```

```python
#changing tuple
#creating tuple
t = (1, 2, 3, 4, [5, 6, 7])
t[2] = 'x'
```

```python
t[4] [1] = 'satish'
print(t)
```

```python
#concatinating tuples

t = (1, 2, 3) + (4, 5, 6)
print(t)
```

```python
#repeat the elements in a tuple for a given number of times using the * operator
t = (('satish', ) *4)
print(t)
```

```python
#tuple deletion
#delete entire tuple using del keyword
#elements in a tuple cannot be deleted

t = (1, 2, 3, 4, 5, 6)

#delete entire tuple
del t
```

```python
#tuple count
t = (1, 2, 3, 1, 3, 3, 4, 1)

#get the frequency of particular element in the tuple
t.count(1)
```

```python
#tuple index
t = (1, 2, 3, 1, 3, 3, 4, 1)
print(t.index(3))   #return index of the first element is equal to 3

#print the index of 1
print(t.index(1))
```

```python
#tuple memebership
#test if an item exists in a tuple or not, using the keyword in
t = (1, 2, 3, 4, 5, 6)
print(1 in t)
```

```python
#tuple length
t = (1, 2, 3, 4, 5, 6)
print(len(t))
```

```python
#tuple sort
t = (4, 5, 1, 2, 3)

new_t = sorted(t)    #take elements in the tuple and return a new sorted list
print(new_t)         #does not sort the tuple itself
```

```python
#get the largest element in a tuple
t = (2, 5, 1, 6, 9)

print(max(t))
```

```python
#get the smallest element in a tuple
print(min(t))
```

```python
#get the sum of elements in the tuple
print(sum(t))
```
